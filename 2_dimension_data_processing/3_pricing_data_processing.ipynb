## Silver

```python
df_bronze = spark.sql(f'SELECT * FROM {catalog}.{bronze_schema}.{data_source}')

display(df_bronze.limit(10))
```

```python
df_bronze.select('month').distinct().show(truncate=False)
```

```python
# Parse 'month' from multiple possible formats
# Applicable when the date is NOT IN DATE TYPE
date_formats = ['yyyy/MM/dd', 'dd/MM/yyyy', 'yyyy-MM-dd', 'dd-MM-yyyy']

df_silver = df_bronze.withColumn(
    'month',
    F.coalesce(
        F.try_to_date(F.col('month'), 'yyyy/MM/dd'),
        F.try_to_date(F.col('month'), 'dd/MM/yyyy'),
        F.try_to_date(F.col('month'), 'yyyy-MM-dd'),
        F.try_to_date(F.col('month'), 'dd-MM-yyyy')
    )
)

display(df_silver.select('month').distinct())
```

## Handling Gross price

```python
df_silver.show(10)
```

```python
from pyspark.sql.types import DoubleType

df_silver = df_silver.withColumn(
    "gross_price",
    F.when(
        F.col("gross_price").cast("string").rlike(r"^-?\d+(\.\d+)?$"),
        F.abs(F.col("gross_price").cast(DoubleType()))
    ).otherwise(F.lit(0.0))
)

df_silver.select('gross_price').distinct().show()
```

```python
# inner join to products table to get the product code
print('gross price rows before join: ', df_silver.count())

df_products = spark.table('fmcg.silver.products')
df_joined = df_silver.join(
    df_products.select('product_id', 'product_code'),
    on='product_id',
    how='inner'
)
df_joined = df_joined.select('product_id', 'product_code', 'month', 'gross_price', 'read_timestamp', 'file_name', 'file_size')

print('gross price rows after join: ', df_joined.count())

display(df_joined.limit(10))
```

```python
df_joined.write\
    .format('delta')\
    .option('delta.enableChangeDataFeed', 'true')\
    .option('mergeSchema', 'true')\
    .mode('overwrite')\
    .saveAsTable(f'{catalog}.{silver_schema}.{data_source}')
```

## Merging Data source with parent


```python
df_gold_price = spark.table('fmcg.gold.sb_dim_gross_price')

df_gold_price.show(5)
```

```python
df_gold_price = (
    df_gold_price
    .withColumn(
        'year',
        F.year('month')
    )
    # 0 = non-zero price,  1 = zero price non zero comes first
    .withColumn(
        'is_zero',
        F.when(
            F.col('gross_price') == 0,
            1
        )
        .otherwise(0)
    )
)

display(df_gold_price.limit(10))
```

```python
w = (
    Window
    .partitionBy('product_code', 'year')
    .orderBy(F.col('is_zero'), F.col('month').desc())
)

df_gold_latest_price = (
    df_gold_price
        .withColumn(
            'rnk',
            F.row_number().over(w)
        )
        .filter(
            F.col('rnk') == 1
        )

)
```

```python
display(df_gold_latest_price.limit(10))
```

```python
# Take required cols

df_gold_latest_price = (
    df_gold_latest_price
    .select(
    'product_code',
    'year',
    'gross_price'
    )
    .withColumnRenamed(
        'gross_price',
        'price_inr'
    )
    .select(
        'product_code',
        'price_inr',
        'year'
    )
)

# change year to string
df_gold_latest_price = df_gold_latest_price.withColumn(
    'year',
    F.col('year').cast('string')
)

display(df_gold_latest_price)
```

```python
delta_table = DeltaTable.forName(spark, 'fmcg.gold.dim_gross_price')

delta_table.alias('target').merge(
    source=df_gold_latest_price.alias('source'),
    condition="target.product_code = source.product_code"
).whenMatchedUpdate(
    set={
        'price_inr': 'source.price_inr',
        'year': 'source.year'
    }
).whenNotMatchedInsert(
    values={
        'product_code': 'source.product_code',
        'price_inr': 'source.price_inr',
        'year': 'source.year'
    }
).execute()
```
